```python
# %% [markdown]
# ## 第二部分：数据读取与可视化

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# 读取数据
df = pd.read_csv('simulation_results.csv')

# 重构三维数据结构
lambda1_values = df['lambda1'].unique()
beta_U_values = df['beta'].unique()
g_values = df['g_value'].unique()

```

```python
plt.figure(figsize=(7.5, 6))

subset = df[df['lambda1'] == lambda1_values[0]] #0.2
# print(subset)

# 重构矩阵
mean_R = subset.pivot(index='beta', columns='g_value', values='mean_R').values
# print(mean_R)
plt.imshow(mean_R, cmap='jet', interpolation='gaussian',
                            extent=[g_values.min(), g_values.max(), 
                                beta_U_values.min(), beta_U_values.max()],
                            origin='lower', aspect='auto',
                            vmin=0, vmax=1)

# # 创建 colorbar 并设置标签及刻度字体大小
# colorbar = plt.colorbar(label='')
# # colorbar.set_label('Colorbar Label', fontsize=16)  # 设置 colorbar 标签字体大小
# colorbar.ax.tick_params(labelsize='18')  # 设置 colorbar 刻度标签字体大小
# # 移动 colorbar 标签到最上方
# colorbar.ax.set_title(r'$\rho_R \left ( \infty \right ) $', fontsize=30, pad=10)  # 设置标签字体大小并调整与colorbar的距离

# # 设置 colorbar 的大小（高度）为原图的一半
# colorbar.ax.set_position([0.8, 0.12, 0.3, 0.65])  # [left, bottom, width, height]

plt.xscale("log")
# 在 y 轴顶部添加统一的指数部分
plt.gca().text(-0.0, 1.0, '1e-03', fontsize='25', verticalalignment='bottom', transform=plt.gca().transAxes)

plt.xlabel(r'$g$', fontsize=40)
plt.ylabel(r'$\beta$', fontsize=40)

# 设置刻度字体大小
plt.xticks([0.01, 0.1, 1], ["$10^{-2}$", "$10^{-1}$", "$10^{0}$"], fontsize='25')
plt.yticks([0.001, 0.0012, 0.0014, 0.0016, 0.0018, 0.0020, 0.0022, 0.0024, 0.0026 ],[r'$1.0$', r'$1.2$', r'$1.4$', r'$1.6$', r'$1.8$', r'$2.0$', r'$2.2$', r'$2.4$', r'$2.6$'], fontsize='25')
plt.savefig('test4_1.pdf', bbox_inches='tight', dpi=300)
plt.show()
```
