```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df_load_beta = pd.read_csv('./peakAI_by_NM_beta.csv', index_col='beta')
df_load_lambda1 = pd.read_csv('./peakAI_by_NM_lambda1.csv', index_col='lambda1')
print(df_load_beta.head())
print('index.name:', df_load_beta.index.name)
print('index:', df_load_beta.index)
print('index.values:', df_load_beta.index.values)
print('columns:',df_load_beta.columns)

beta_values = df_load_beta.index.values
lambda_values = df_load_lambda1.index.values

```

```python
# Plotting
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 或其他字体名称
plt.rcParams['axes.unicode_minus'] = False
# 绘图

plt.figure(figsize=(9, 6))
styles = [
    # {'label': 'N=10,M=20', 'color': '#0466c8', 'marker': 'o'},
    # {'label': 'N=12,M=18', 'color': '#009999', 'marker': 'v'},
    # {'label': 'N=15,M=15', 'color': '#BF3147', 'marker': 's'},
    # {'label': 'N=18,M=12', 'color': '#FF8000', 'marker': 'P'},
    # {'label': 'N=20,M=10', 'color': '#4E9980', 'marker': 'D'},
    {'label': 'N=10, M=20', 'color': '#735C8B', 'marker': 'o'},
    {'label': 'N=12, M=18', 'color': '#DAA57C', 'marker': 'v'},
    {'label': 'N=15, M=15', 'color': '#E93A32', 'marker': 's'},
    {'label': 'N=18, M=12', 'color': '#526BA4', 'marker': 'P'},
    {'label': 'N=20, M=10', 'color': '#4E9980', 'marker': 'D'},
]

for i, style in enumerate(styles):
    plt.plot(beta_values,
             df_load_beta[style['label']],
             label=style['label'],
             color=style['color'],
             marker=style['marker'],
             markevery=2,
             linewidth=3,
             linestyle='--',
             markersize=10)

legend = plt.legend(
    loc='lower right',
    fontsize='17',
    handletextpad=0.5,    # 图标与文本的间距
    borderpad=0.4,          # 图例内边距
    handlelength=3.5,    # 统一图标宽度
)
legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1)
    
# 启用四面刻度
plt.tick_params(
    axis='both',      # 同时控制 x 和 y 轴
    which='both',     # 同时控制主刻度和次刻度
    top=True,         # 显示顶部刻度
    right=True,       # 显示右侧刻度
    labeltop=False,   # 不显示顶部刻度标签（避免重叠）
    labelright=False,  # 不显示右侧刻度标签（避免重叠）
    direction='in',   # 刻度朝内
)


plt.xlabel(r'$\beta$', fontsize=30)
# plt.xticks([0.000, 0.005, 0.010, 0.015, 0.020, 0.025, 0.030],[r'$0$', r'$0.005$', r'$0.010$', r'$0.015$', r'$0.020$', r'$0.025$', r'$0.030$'], fontsize=20)
# plt.xlim(-0.001, 0.031)
plt.xticks([0.000, 0.002, 0.004, 0.006, 0.008, 0.010, 0.012],[ r'$0$', r'$0.002$', r'$0.004$', r'$0.006$', r'$0.008$', r'$0.010$', r'$0.012$' ],fontsize=20)
plt.xlim(-0.0005, 0.013)


plt.ylabel('感染密度峰值', fontsize=28)
plt.yticks([0.0, 0.2, 0.4, 0.6, 0.8],[r'$0.0$', r'$0.2$', r'$0.4$', r'$0.6$', r'$0.8$'],fontsize=20)
plt.ylim(-0.08, 0.88)

plt.savefig('graduation_test7_1.pdf', bbox_inches='tight', dpi=300)
plt.show()
```
