```python
import numpy as np

# 假设有 3 个居住点，每个点有 5 种状态
states = np.array([[0.1, 0.2, 0.3, 0.2, 0.2],
                   [0.4, 0.1, 0.2, 0.1, 0.2],
                   [0.3, 0.3, 0.1, 0.2, 0.1]])

# 各居住点人口
n = np.array([100, 200, 300])  # (3,)


# 计算结果
result = states * n[:, None]  # (3,5)   n[:, None]扩展为[[100,100,100,100,100],[200,200,200,200,200],[300,300,300,300,300]]

print("states:\n", states)
print("n[:, None]:\n", n[:, None])
print("result:\n", result)
```

```python
import numpy as np

# 定义二分图的边 (U和V的顶点编号分两组)
edges = [(0, 2), (0, 4), (1, 3)]  # u0=0, u1=1; v0=2, v1=3, v2=4

# 初始化邻接矩阵 (5x5, 因为 |U|+|V|=2+3=5)
A = np.zeros((5, 5), dtype=int)

# 填充边
for u, v in edges:
    A[u][v] = 1
    A[v][u] = 1  # 无向图需对称

print("邻接矩阵:\n", A)
```

```python
# 设置颜色和标签
state_colors = {
    'US': 'blue',
    'AS': 'green',
    'AI': 'red',
    'UR': 'purple',
    'AR': 'orange'
}


state_labels = ['US', 'AS', 'AI', 'UR', 'AR']

#这段代码是 Python 中用于 同时遍历多个列表 的常用技巧
'''
  ---- zip(state_labels, state_colors.values()):将两个列表按位置配对，生成一个可迭代的元组序列：
       [('US', 'blue'), ('AS', 'green'), ('AI', 'red'), ('UR', 'purple'), ('AR', 'orange')]

  ---- enumerate(zip(state_labels, state_colors.values())):将上述元组序列按位置索引,生成一个enumerate对象.
  [
    (0, ('US', 'blue')),
    (1, ('AS', 'green')),
    (2, ('AI', 'red')),
    (3, ('UR', 'purple')),
    (4, ('AR', 'orange'))
  ]
'''


for i, (label, color) in enumerate(zip(state_labels, state_colors.values())):  
    print(i, label, color)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 50)
y = np.sin(x)

plt.plot(x, y, 
         linestyle='-',  # 实线
         marker='o',     # 圆圈标记
         markevery=5,    # 每5个数据点显示一个标记
         markersize=8,
         mfc = 'white',
         label='markevery=5')

plt.legend()
plt.grid(True)
plt.show()
```

```python
# 第一步：绘制线条（无标记）
plt.plot(x, y, linestyle='-', color='blue', label='Line')

# 第二步：绘制标记（无线条）
plt.plot(x[::5], y[::5],  # 每隔5个点取一个标记
         linestyle='',     # 无线条
         marker='o', 
         color='red',
         label='Markers')

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 30)
y = np.sin(x)

linestyles = ['-', '--', ':', '-.', (0, (5, 2))]
labels = ['Solid', 'Dashed', 'Dotted', 'Dashdot', 'Custom Dashed']

plt.figure(figsize=(10, 5))
for ls, label in zip(linestyles, labels):
    plt.plot(x, y, 
             linestyle=ls, 
             marker='o', 
             markevery=3,  # 每3个点显示一个标记
             label=f'{label} + markevery')

plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

np.random.seed(2025)
w1 = np.random.rand(3, 3)  # 第一次运行
print("第一次生成的 w:\n", w1)

np.random.seed(2025)       # 重置种子
w2 = np.random.rand(3, 3)  # 第二次运行
print("第二次生成的 w:\n", w2)

print("两次结果是否相同:", np.allclose(w1, w2))  # 输出 True
```
