```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# read the data
data1_0= pd.read_csv('./exp1_0_peak.csv', index_col=0)
print(data1_0.head())
# print(data1_0['0.01'])
lambda_values1 = data1_0.index.values
# print(lambda_values)

data1_50= pd.read_csv('./exp1_50_peak.csv', index_col=0)
lambda_values2 = data1_50.index.values

data1_60= pd.read_csv('./exp1_60_peak.csv', index_col=0)
lambda_values3 = data1_60.index.values
```

```python
# 绘图

param_combinations = [
    {'epsilon': '0.01', 'label': r'$ \epsilon_0 $ = 0.01', 'color': '#0466c8', 'marker': '^','color2': "#063565",},
    {'epsilon': '0.3',  'label': r'$ \epsilon_0 $ = 0.3',  'color': '#BF3147', 'marker': 'X','color2': "#811727",},
    {'epsilon': '0.9',  'label': r'$ \epsilon_0 $ = 0.9',  'color': '#4E9980', 'marker': 'o','color2': "#05553A",},
]



plt.figure(figsize=(9, 6))

for comb in param_combinations:
    epsilon = comb['epsilon']
    label = comb['label']
    # print(data1_0[epsilon])
    plt.scatter(
        lambda_values1,
        data1_0[epsilon],
        label=label,
        color=comb['color'],
        marker=comb['marker'],
        edgecolors=comb['color2'],     # 白色填充+黑边
        s=140,                  # 控制点大小
        linewidths=0.5,
        zorder=3
    )

# 图例设置
legend = plt.legend(
    loc='upper right',
    fontsize=20,
    handletextpad=0.3,
    borderpad=0.4,
    handlelength=1.8,
)
legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1)

# 坐标轴刻度设置
plt.tick_params(
    axis='both',
    which='both',
    top=True,
    right=True,
    labeltop=False,
    labelright=False,
    direction='in',
)

plt.xlabel(r'$\lambda$', fontsize=30)
plt.xticks([0.3, 0.4, 0.5, 0.6, 0.7], [r'$0.3$', r'$0.4$', r'$0.5$', r'$0.6$', r'$0.7$'], fontsize=20)
plt.xlim(0.28, 0.72)

plt.ylabel('peak infection density', fontsize=28)
plt.yticks([0.46, 0.48, 0.5, 0.52, 0.54], [r'$0.46$', r'$0.48$', r'$0.50$', r'$0.52$', r'$0.54$'], fontsize=20)
plt.ylim(0.452, 0.548)

# plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.5, zorder=0)

plt.savefig('test5_1.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
