```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read the data 
data1 = pd.read_csv('./exp2_60_alpha01_09.csv', index_col='lambda')
lambda_values = data1.index.values
label_1 = data1.columns.values

data2 = pd.read_csv('./exp2_60_alpha03_06.csv', index_col='lambda')
lambda_values2 = data2.index.values
label_2 = data2.columns.values

print('label_1:',label_1[0])
print('lambda_values:',lambda_values[::3])
data1.head()
data1.iloc[:, 0].head()
data_test = data1.iloc[::3].head()  # 每三个取一个数据
print('data1.iloc[::3].head() :',data_test)
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 或其他字体名称
plt.rcParams['axes.unicode_minus'] = False
# 绘图

plt.figure(figsize=(9, 6))

styles = [
    {'color': '#526BA4', 'marker': '^','color2': "#063565",},
    {'color': '#4E9980', 'marker': 'o','color2': "#05553A",},
    {'color': '#E93A32', 'marker': 'p','color2': "#811727",},
    {'color': '#735C8B', 'marker': 'v','color2': "#402334",},
]

for i, style in enumerate(styles):
    plt.scatter(
        lambda_values[::3],
        data1.iloc[::3, i],
        label=label_1[i],
        color=style['color'],
        marker=style['marker'],
        s=150,                      # 控制点大小（相当于原markersize=10的放大版）
        edgecolors=style['color2'],         # 黑色边缘
        linewidths=0.5,             # 边缘线宽（相当于原markeredgewidth）
        zorder=3                    # 确保散点在上层
    )

legend = plt.legend(
    loc='upper right',
    fontsize='18',
    handletextpad=0.5,
    borderpad=0.4,
    handlelength=3.5,
)
legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1)
    
# 刻度设置（保持不变）
plt.tick_params(
    axis='both',
    which='both',
    top=True,
    right=True,
    labeltop=False,
    labelright=False,
    direction='in',
)

# 坐标轴标签和范围（保持不变）
plt.xlabel(r'$\lambda$', fontsize=30)
plt.xticks([0.4, 0.5, 0.6, 0.7, 0.8], 
           [r'$0.4$', r'$0.5$', r'$0.6$', r'$0.7$', r'$0.8$'], 
           fontsize=20)
plt.xlim(0.38, 0.82)

plt.ylabel('感染密度峰值', fontsize=28)
plt.yticks([0.45, 0.46, 0.47, 0.48, 0.49, 0.50, 0.51, 0.52],
           [r'$0.45$', r'$0.46$', r'$0.47$', r'$0.48$', r'$0.49$', r'$0.50$', r'$0.51$', r'$0.52$'],
           fontsize=20)
plt.ylim(0.447, 0.523)

plt.savefig('graduation_test6_1.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
