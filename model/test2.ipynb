```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 或其他字体名称
plt.rcParams['axes.unicode_minus'] = False
# 绘图
# ============================================
# 1. 读取两个模型的峰值数据
# ============================================
cps_df = pd.read_csv("CPS_peak_vs_lambda.csv")
mir_df = pd.read_csv("MIR_peak_vs_lambda.csv")

# ============================================
# 2. 绘制对比曲线（无误差棒）
# ============================================
plt.figure(figsize=(9, 6))

# CPS 模型：红色圆点连线
plt.plot(cps_df["lambda"], cps_df["peak_mean"],
         'o-', color='#BF3147', linewidth=3.5, mfc='white',markersize=10, markevery = 3,
         label='CPS Model')

# MIR 传统 UAU-SIR 模型：蓝色方块连线
plt.plot(mir_df["lambda"], mir_df["peak_mean"],
         's-', color='#526BA4', linewidth=3.5, mfc='white',markersize=10, markevery = 3,
         label='MIR Model')

# ============================================
# 3. 图表装饰
# ============================================

plt.tick_params(
    axis='both',      # 同时控制 x 和 y 轴
    which='both',     # 同时控制主刻度和次刻度
    top=True,         # 显示顶部刻度
    right=True,       # 显示右侧刻度
    labeltop=False,   # 不显示顶部刻度标签（避免重叠）
    labelright=False,  # 不显示右侧刻度标签（避免重叠）
    direction='in',   # 刻度朝内
)

legend = plt.legend(
    loc=(0.68, 0.46),
    fontsize='18',
    handletextpad=0.3,    # 图标与文本的间距
    #columnspacing=1,      # 列间距（多列时有效）
    borderpad=0.4,          # 图例内边距
    ncol=1,              # 分1列
    handlelength=2.5,    # 统一图标宽度
    columnspacing=0.5,     # 缩小列间距（默认2.0）
)

legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1)

plt.xlabel(r'$\lambda$',fontsize='30')
plt.ylabel('感染密度峰值',fontsize='30')
plt.xticks(fontsize='24')
plt.xlim(0.39, 0.81)  # 设置x轴范围
plt.yticks(fontsize='24')
plt.ylim(0.25, 0.40)  # 设置y轴范围



plt.tight_layout()

# 保存图片（高清）
plt.savefig('graduation_compare_peak.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
